```python
from util import fisherztrans,stat_m_e,pair_test,sig,color_inv_alpha
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import scipy.stats as stats
from statsmodels.stats.multitest import multipletests
import seaborn as sns
from faceprf_figure_function import set_figure,set_ax,myviolinplot_ind,myviolinplot_pair
import pingouin

cms=np.array([[255, 195, 73],[145, 115, 185]])/255 

nsubj = 8
npos = 16
task = ['Digit Task','Face Task']
roi_labels = ['V1','V2','V3','hV4','IOG','pFus','mFus']
nroi = len(roi_labels)
ntask = len(task)
# distance: near vs. far
num_nf=[[0,4],[5,9]]
nf = ['Near','Far']
# roi: early visual cortex (V1-V3) - face-selective region
num_vf=[[0,3],[4,7]]
vf = ['V1-V3','OFA,FFA-1,FFA-2']
nsample2 = nsubj * np.max(np.diff(num_nf)) * np.max(np.diff(num_vf))
nvertex = 100

data_path = "../../fmri_analysis/expectedoutput/"
```

### |df|

```python
dfnorm_d_subj = np.load(data_path+'dfnorm.npz',allow_pickle=True)['dfnorm_d_subj'] # d x subj x roi x task

# average accuracy for near/far distance
dfnorm_nf_subj = np.zeros([4*nsubj,nroi,ntask,len(nf)])*np.nan # 4*8samples x roi x task x nf
for task_i in range(ntask):
    for roi_i in range(nroi):
        for i in range(len(num_nf)):
            data = dfnorm_d_subj[num_nf[i][0]:num_nf[i][1],:,roi_i,task_i]
            dfnorm_nf_subj[:,roi_i,task_i,i] = data.reshape([data.shape[0]*data.shape[1]])
dfnorm_nf_subj[:,:,[0,1],:] = dfnorm_nf_subj[:,:,[1,0],:] # 4*8samples x roi x task x nf

# far distance condition
i = 1 # nf=1(far)

# test
ps0,sts = pair_test(dfnorm_nf_subj[:,:,:,i],method='wilcoxon',correction='none')
ps,sts = pair_test(dfnorm_nf_subj[:,:,:,i],method='wilcoxon',correction='bonferroni')
print('p( without correction):{}'.format([round(p,4) for p in ps0]))
print('p(correction):{}'.format([round(p,4) for p in ps]))    

# plot
ylims = [[0,50,10],[0,50,10]]
pys = [48,47]
set_figure()
fig,axes = plt.subplots(1,1,figsize=(7,4))
set_ax(axes)
myviolinplot_pair(axes,dfnorm_nf_subj[:,:,:,i],'mean','sem',ps,pys)
axes.set_ylabel('Signal Separation\n(% BOLD change)')
axes.set_xticklabels(roi_labels)
axes.set_yticks(np.arange(ylims[i][0],ylims[i][1]*1.01,ylims[i][2]))
axes.set_xlim([0.3,7.7])
axes.set_ylim(ylims[i][:2])
fig.savefig('../figures/fig5-dfnorm.pdf', bbox_inches='tight')
```

### Variance


```python
v_d_subj = np.load(data_path+'meanvariance.npz',allow_pickle=True)['v_d_subj'] # d x subj x roi x task

# average accuracy for near/far distance
v_nf_subj = np.zeros([4*nsubj,nroi,ntask,len(nf)])*np.nan # 4*8samples x roi x task x nf
for task_i in range(ntask):
    for roi_i in range(nroi):
        for i in range(len(num_nf)):
            data = v_d_subj[num_nf[i][0]:num_nf[i][1],:,roi_i,task_i]
            v_nf_subj[:,roi_i,task_i,i] = data.reshape([data.shape[0]*data.shape[1]])
v_nf_subj[:,:,[0,1],:] = v_nf_subj[:,:,[1,0],:] # 4*8samples x roi x task x nf


# far distance condition
i = 1 # nf=1(far)

# test
ps0,sts = pair_test(v_nf_subj[:,:,:,i],method='wilcoxon',correction='none')
ps,sts = pair_test(v_nf_subj[:,:,:,i],method='wilcoxon',correction='bonferroni')
print('p( without correction):{}'.format([round(p,4) for p in ps0]))
print('p(correction):{}'.format([round(p,4) for p in ps]))    

# plot
ylims = [[0,45,10],[0,45,10]]
pys = [43.5,43]
set_figure()
fig,axes = plt.subplots(1,1,figsize=(7.2,4))
set_ax(axes)
myviolinplot_pair(axes,v_nf_subj[:,:,:,i],'mean','sem',ps,pys)
axes.set_ylabel('Mean Variance'+'\n(% '+ 'BOLD change'+'$^2$'+')')
axes.set_xticklabels(roi_labels)
axes.set_yticks(np.arange(ylims[i][0],ylims[i][1]*1.01,ylims[i][2]))
axes.set_xlim([0.3,7.7])
axes.set_ylim(ylims[i][:2])
fig.savefig('../figures/fig5-meanvariance.pdf', bbox_inches='tight')
```

### Signal rotation

```python
cm = [[0.5,0.5,0.5],'#227093']
mks = ['s','x']
lws = [0.4,1]

srangle_d_subj = np.load(data_path+'srangle.npz',allow_pickle=True)['srangle_d_subj'] # d x subj x roi
srangle_d_subj_ctl = np.load(data_path+'srangle_ctl.npz',allow_pickle=True)['srangle_ctl'] # task x d x subj x roi

# average srangle for near/far distance
srangle_nf_subj = np.zeros([4*nsubj,nroi,len(nf)])*np.nan # 4*8samples x roi x nf
for roi_i in range(nroi):
    for i in range(len(num_nf)):
        data = srangle_d_subj[num_nf[i][0]:num_nf[i][1],:,roi_i]
        srangle_nf_subj[:,roi_i,i] = data.reshape([data.shape[0]*data.shape[1]])

srangle_nf_subj_ctl = np.zeros([ntask,4*nsubj,nroi,len(nf)])*np.nan # task x 4*8samples x roi x nf
for task_i in range(ntask):
    for roi_i in range(nroi):
        for i in range(len(num_nf)):
            data = srangle_d_subj_ctl[task_i,num_nf[i][0]:num_nf[i][1],:,roi_i]
            srangle_nf_subj_ctl[task_i,:,roi_i,i] = data.reshape([data.shape[0]*data.shape[1]])
srangle_nf_subj_ctl[[0,1],:] = srangle_nf_subj_ctl[[1,0],:]

# far distance condition
nf_i = 1 # nf_i=1(far)

# plot
ylims = [[0,150,30],[0,150,30]]
set_figure()
fig,ax = plt.subplots(1,1,figsize=(7,4))
set_ax(ax)


nconditions = 2
sides = ['low','high']
xgap = 1; nxl = len(roi_labels); w = 1.8/nconditions*xgap
x = np.linspace(xgap,xgap*nxl,nxl)
data_m,data_e = np.zeros([nxl,nconditions]),np.zeros([2,nxl,nconditions])

# left: baseline
data = np.vstack([srangle_nf_subj_ctl[0,...,nf_i],srangle_nf_subj_ctl[1,...,nf_i]])
[data_m[:,0],data_e[:,:,0],_] = stat_m_e(data,'mean','sem')
vp1 = ax.violinplot(data, positions = x-0.04, vert = True, widths = w*0.85, 
                points = 100, bw_method = 'scott', side = sides[0],
            showmeans = False, showextrema = False, showmedians = False, quantiles = None)
for b1 in vp1['bodies']:
    b1.set_edgecolor("none")
    b1.set_facecolor(cm[0])
    b1.set_alpha(0.2)

vp1 = ax.violinplot(data, positions = x-0.04, vert = True, widths = w*0.85, 
                    points = 100, bw_method = 'scott', side = sides[0],
                showmeans = False, showextrema = False, showmedians = False, quantiles = None)
for b1 in vp1['bodies']:
    b1.set_edgecolor(cm[0])
    b1.set_facecolor("none")
    b1.set_alpha(0.7)

ax.errorbar(x-0.05,data_m[:,0],data_e[:,:,0],
                    color='k',marker='',linestyle='',elinewidth=1.2)
ax.plot([x-0.12,x-0.05],[data_m[:,0],data_m[:,0]],
                    color='k',linewidth=1.2)

for task_i in range(ntask):
    jitter = np.random.uniform(0,w/8,srangle_nf_subj_ctl[task_i,...,nf_i].shape)
    ax.scatter(np.zeros_like(srangle_nf_subj_ctl[task_i,...,nf_i])+x-(0.2*(2-task_i)+jitter)+0.06, srangle_nf_subj_ctl[task_i,...,nf_i], 
            color = cm[0], alpha = 0.6, s = 35, edgecolors = 'white', linewidths=lws[task_i], marker = mks[task_i])
    #

ax.set_xticks(x)    

# right: task effect
data = srangle_nf_subj[...,nf_i]
[data_m[:,1],data_e[:,:,1],_] = stat_m_e(data,'mean','sem')

vp1 = ax.violinplot(data, positions = x+0.04, vert = True, widths = w*0.85, 
                points = 100, bw_method = 'scott', side = sides[1],
            showmeans = False, showextrema = False, showmedians = False, quantiles = None)
for b1 in vp1['bodies']:
    b1.set_edgecolor("none")
    b1.set_facecolor(cm[1])
    b1.set_alpha(0.2)

vp1 = ax.violinplot(data, positions = x+0.04, vert = True, widths = w*0.85, 
                    points = 100, bw_method = 'scott', side = sides[1],
                showmeans = False, showextrema = False, showmedians = False, quantiles = None)
for b1 in vp1['bodies']:
    b1.set_edgecolor(cm[1])
    b1.set_facecolor("none")
    b1.set_alpha(0.7)

ax.errorbar(x+0.05,data_m[:,1],data_e[:,:,1],
                    color='k',marker='',linestyle='',elinewidth=1.2)
ax.plot([x+0.12,x+0.05],[data_m[:,1],data_m[:,1]],
                    color='k',linewidth=1.2)

jitter = np.random.uniform(0,w/8,data.shape)
ax.scatter(np.zeros_like(data)+x+(0.15+jitter), data, 
        color = cm[1], alpha = 0.6, s = 35, edgecolors = 'white',linewidths=0.2)
ax.set_xticks(x)


ax.set_ylabel('Signal Rotation Angle ($\degree$)')
ax.set_xticklabels(roi_labels)
ax.set_yticks(np.arange(ylims[nf_i][0],ylims[nf_i][1]*1.01,ylims[nf_i][2]))
ax.set_xlim([0.3,7.7])
ax.set_ylim(ylims[nf_i][:2])

fig.savefig('../figures/fig5-srangle.pdf', bbox_inches='tight')
```
