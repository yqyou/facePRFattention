#### 2.1 Fano Factor

The ratio of the variance to the mean is often called the **Fano factor** and is a measure of the variability in the number of spikes in relation to the mean number of spikes:

$$
FF(T)≡\frac{σ_{N(0,T)}^2}{m_{N(0,T)}}
$$
$$
FF_i = \frac{v_i}{b_i}
$$
$$
FF^n = median(FF_i)
$$


```python
# calculate
FF_p_all = np.zeros((npos,nsubj,len(roi_labels),len(tasks)))*np.nan # [npos(s1),nsubj,nroi,ntask]
FFv_all = np.zeros((nvoxel,npos,nsubj,len(roi_labels),len(tasks)))*np.nan # [nvoxel,npos(s1),nsubj,nroi,ntask]
for task_i in range(len(tasks)):
    for roi_i in range(len(roi_labels)):
        for subj_i in range(len(subjs)):
            stimposicond = stimposicond_all[task_i,subj_i,:]
            # beta for each position (trial-by-trial)
            beta = np.zeros((nvoxel,ntrialperpos,npos))*np.nan # voxel x trial x position
            beta_mean = np.zeros((nvoxel,npos))*np.nan
            for pos in range(npos):
                beta[:,:,pos] = beta_all[task_i,roi_i,subj_i,:,stimposicond==pos+1].T
                beta_mean[:,pos] = np.nanmean(beta[:,:,pos],axis=1)
                v = np.var(beta[:,:,pos],axis=1) # 100
                f = np.mat(beta_mean[:,pos]) # 1 x 100
                # 1 x 100
                # abs
                FF_n = np.multiply(v,abs(1/f))
                FFv_all[:,pos,subj_i,roi_i,task_i] = np.array(FF_n)
                FF_p_all[pos,subj_i,roi_i,task_i] = np.nanmedian(np.array(FF_n))

# concat data across position
FF_all = np.zeros((npos*nsubj,len(roi_labels),len(tasks)))*np.nan
for task_i in range(len(tasks)):
    for roi_i in range(len(roi_labels)):
        c = np.array([])
        for subj_i in range(len(subjs)):
            c = np.append(c,FF_p_all[:,subj_i,roi_i,task_i])
        FF_all[:,roi_i,task_i] = c

np.savez('../expectedoutput/FF',FF_p_all=FF_p_all,FF_all=FF_all,FFv_all=FFv_all)           
```

#### 3.1 I_bc

```python
def cal_lfi(beta_mean_cond,beta_cond,s1,s2,log=False):
    # ds
    dr = s1//4-s2//4
    dc = s1%4-s2%4
    ds = ds0 *  (dr**2+dc**2)**0.5
    # df & |dfnorm|
    df = np.mat(beta_mean_cond[:,s1] - beta_mean_cond[:,s2])
    dfnorm = np.linalg.norm(df,axis=1)
    # eigenvector & eigenvalue
    q1 = np.cov(beta_cond[:,:,s1])
    q2 = np.cov(beta_cond[:,:,s2])
    q = (q1 + q2)/2 # 100 x 100
    w,v = np.linalg.eigh(q)
    ullambda = np.nanmean(w)
    htlambda = w/ullambda
    # lfi
    lfi = dfnorm**2/ullambda*np.nansum([(df/dfnorm@v[:,i])**2/htlambda[i] for i in range(len(htlambda))])/(ds)**2
    #lfi = df@np.linalg.inv(q)@df.T/(ds)**2
    if log:
        lfi = np.log(lfi)
    return lfi,ds
```


$$I=\frac{df^T*S^{-1}*df}{ds^2}$$
$$\hat I_{bc}=I*\frac{2T-N-3}{2T-2}-\frac{2N}{Tds^2}$$

$S$ is the empirical covariance for each presented stimulus:
$$
S_1 = \frac{1}{T-1} \displaystyle\sum^T(\bold r - \bold f)^T(\bold r - \bold f)
$$

### 4 LFI: four mechanisms

```python
def cal_lfi_param(beta_mean_cond,beta_cond,s1,s2):
    # ds
    dr = s1//4-s2//4
    dc = s1%4-s2%4
    ds = ds0 *  (dr**2+dc**2)**0.5
    # df & |dfnorm|
    df = np.mat(beta_mean_cond[:,s1] - beta_mean_cond[:,s2])

    dfnorm = np.linalg.norm(df,axis=1)
    # eigenvector & eigenvalue
    q1 = np.cov(beta_cond[:,:,s1])
    q2 = np.cov(beta_cond[:,:,s2])
    q = (q1 + q2)/2 # 100 x 100

    w,v = np.linalg.eigh(q)
    ullambda = np.nanmean(w)
    htlambda = w/ullambda
    return ds,df,dfnorm,v,ullambda,htlambda

def cal_lfi(ds,dfnorm,ullambda,df_dfnorm,v,htlambda,bc=True,ntrialperpos=ntrialperpos, nvoxel=nvoxel):
    I = 1/(ds)**2*dfnorm**2/ullambda*np.nansum([(df_dfnorm@v[:,i])**2/htlambda[i] for i in range(len(htlambda))])
    if bc:
        I_bc = I*(2*ntrialperpos-nvoxel-3)/(2*ntrialperpos-2)-(2*nvoxel)/(ntrialperpos*(ds)**2)
        return I_bc
    else:
        return I

def cal_lfi_log(ds,dfnorm,ullambda,df_dfnorm,v,htlambda,bc=True,ntrialperpos=ntrialperpos, nvoxel=nvoxel):
    I = cal_lfi(ds,dfnorm,ullambda,df_dfnorm,v,htlambda,bc,ntrialperpos, nvoxel)
    return np.log(I)
```

```python
def runlfi(pi,sfl_n,task_i,sfl_i,save=True):
    # sfl_type: 'ori' or 'sfld'
    LFI_bc_ij = np.full((sfl_n,npos,npos,nsubj,len(roi_labels)),np.nan) # [nshuffle,npos(s1),npos(s2),nsubj,nroi]
    for roi_i in range(len(roi_labels)):
        for subj_i in range(len(subjs)):
            print('pi{},roi{},subj{}'.format(pi,roi_i,subj_i))
            
            stimposicond0 = stimposicond_all[task_i,subj_i,:].copy() # trial
            beta0 = np.zeros((nvoxel,ntrialperpos,npos))*np.nan # voxel x trial x position
            for pos in range(npos):
                beta0[:,:,pos] = beta_all[task_i,roi_i,subj_i,:,stimposicond0==pos+1].T  
            
            # shuffle
            for i in tqdm(range(sfl_n)):
                if sfl_i == 0:

                    idx = np.array([np.random.permutation(ntrialperpos) for _ in range(npos)]).T
                    beta_sfl = np.take_along_axis(beta0, idx[None,:,:], axis=1) 
                elif sfl_i == 1:

                    idx = np.argsort(np.random.rand(nvoxel, ntrialperpos, npos), axis=1)
                    beta_sfl = np.take_along_axis(beta0, idx, axis=1)

                # bias-corrected LFI
                beta_sfl_mean = np.nanmean(beta_sfl,axis=1)

                for s1 in range(npos-1):
                    for s2 in range(s1+1,npos):
                        [LFI_ij,ds] = cal_lfi(beta_sfl_mean,beta_sfl,s1,s2)
                        LFI_bc = LFI_ij*(2*ntrialperpos-nvoxel-3)/(2*ntrialperpos-2)-(2*nvoxel)/(ntrialperpos*(ds)**2)
                        LFI_bc_ij[i,[s1,s2],[s2,s1],subj_i,roi_i] = LFI_bc
    return LFI_bc_ij

def parallel_lfi_shuffle(sfl_total,task_i,sfl_i,sfl_types,save=True):
    npall = 20
    sfl_n = int(sfl_total/npall)
    results = Parallel(n_jobs=-1)(delayed(runlfi)(
        pi=pi,
        sfl_n=sfl_n,
        task_i=task_i,
        sfl_i=sfl_i,
        save=save
        )
        for pi in range(npall))
    lfi_ij = np.concatenate(results, axis=0)

    # save                        
    if save: np.savez_compressed('../expectedoutput/lfi-bc_shuffle_'+sfl_types[sfl_i]+'_'+tasks[task_i],lfi_ij=lfi_ij)  

    return lfi_ij
  
def cal_lfi(beta_mean_cond,beta_cond,s1,s2,log=False):
    # ds
    dr = s1//4-s2//4
    dc = s1%4-s2%4
    ds = ds0 *  (dr**2+dc**2)**0.5
    # df & |dfnorm|
    df = beta_mean_cond[:,s1] - beta_mean_cond[:,s2]
    dfnorm = np.linalg.norm(df)
    # eigenvector & eigenvalue
    q1 = np.cov(beta_cond[:,:,s1])
    q2 = np.cov(beta_cond[:,:,s2])
    q = (q1 + q2)/2 # 100 x 100
    w,v = np.linalg.eigh(q)
    ullambda = np.nanmean(w)
    htlambda = w/ullambda
    # lfi
    proj = (df/dfnorm) @ v
    lfi = (dfnorm**2/ullambda) * np.nansum(np.square(proj)/ htlambda)/(ds**2)
    if log:
        lfi = np.log(lfi)
    return lfi,ds
```

```python
from scipy.stats import multivariate_normal
from concurrent.futures import ThreadPoolExecutor, as_completed

def simulate_data(beta_all, stimposicond_all, ntrialperpos_sim = 80, rng=None):

    rng = np.random.default_rng() if rng is None else rng

    ntask,nroi,nsubj,nvoxel,_ = beta_all.shape

    ntrial_sim = ntrialperpos_sim * npos

    # generate stimposicond_sim_all
    stimposicond_sim = np.repeat(np.arange(1,npos+1),ntrialperpos_sim,)
    stimposicond_sim_all = np.tile(stimposicond_sim, (ntask,nsubj,1))
    
    beta_sim_all = np.full((ntask,nroi,nsubj,nvoxel,ntrial_sim), np.nan)

    for task_i in range(ntask):
        for roi_i in range(nroi):
            for subj_i in range(nsubj):
                beta_raw = beta_all[task_i,roi_i,subj_i]
                stim_raw = stimposicond_all[task_i,subj_i]

                stim_sim = stimposicond_sim_all[task_i,subj_i]
      
                for s in range(npos):
                    idx_raw = stim_raw == s + 1
                    X_raw = beta_raw[:, idx_raw]

                    mean0 = np.nanmean(X_raw, axis=1)
                    cov0 = np.cov(X_raw)
                    cov0 += np.eye(nvoxel) * 1e-6

                    idx_sim = stim_sim == s + 1

                    beta_sim_all[task_i,roi_i,subj_i,:,idx_sim] = rng.multivariate_normal(mean=mean0, cov=cov0, size=ntrialperpos_sim)
        
    return beta_sim_all, stimposicond_sim_all


def cal_lfi_param_cov(beta_mean_cond,beta_cov,s1,s2):
    # ds
    dr = s1//4-s2//4
    dc = s1%4-s2%4
    ds = ds0 *  (dr**2+dc**2)**0.5
    # df & |dfnorm|
    df = np.mat(beta_mean_cond[:,s1] - beta_mean_cond[:,s2])

    dfnorm = np.linalg.norm(df,axis=1)
    # eigenvector & eigenvalue
    q1 = beta_cov[:,:,s1]
    q2 = beta_cov[:,:,s2]
    q = (q1 + q2)/2 # 100 x 100

    w,v = np.linalg.eigh(q)
    ullambda = np.nanmean(w)
    htlambda = w/ullambda
    return ds,df,dfnorm,v,ullambda,htlambda

def cal_lfi(ds,dfnorm,ullambda,df_dfnorm,v,htlambda,bc=True,ntrialperpos=ntrialperpos, nvoxel=nvoxel):
    I = 1/(ds)**2*dfnorm**2/ullambda*np.nansum([(df_dfnorm@v[:,i])**2/htlambda[i] for i in range(len(htlambda))])
    if bc:
        return cal_lfi_bc(I, ntrialperpos, nvoxel, ds)
    else:
        return I

def cal_lfi_bc(I, ntrialperpos, nvoxel, ds):
    return I*(2*ntrialperpos-nvoxel-3)/(2*ntrialperpos-2)-(2*nvoxel)/(ntrialperpos*(ds)**2)

def cal_lfi_log_fast(ds,dfnorm,ullambda,df_dfnorm,v,htlambda,bc=True,ntrialperpos=ntrialperpos, nvoxel=nvoxel):
    projection = np.asarray(df_dfnorm) @ v
    I = (dfnorm*dfnorm / (ds * ds) / ullambda) * np.nansum(projection * projection / htlambda)
    #I = 1/(ds)**2*dfnorm**2/ullambda*np.nansum([(df_dfnorm@v[:,i])**2/htlambda[i] for i in range(len(htlambda))])
    if bc:
        I_bc = I*(2*ntrialperpos-nvoxel-3)/(2*ntrialperpos-2)-(2*nvoxel)/(ntrialperpos*(ds)**2)
        return np.log(I_bc)
    else:
        return np.log(I)

def cal_lfi_log(ds,dfnorm,ullambda,df_dfnorm,v,htlambda,bc=True,ntrialperpos=ntrialperpos, nvoxel=nvoxel):
    I = cal_lfi(ds,dfnorm,ullambda,df_dfnorm,v,htlambda,bc,ntrialperpos, nvoxel)
    return np.log(I)

def lfi_mechanism_task(beta_all,stimposicond_all):
    _,nroi,nsubj,nvoxel,ntrial = beta_all.shape
    ntrialperpos = int(ntrial/npos)
    Ilog_stw_ij_all = np.zeros((npos,npos,nsubj,len(roi_labels),5))*np.nan # [npos(s1),npos(s2),nsubj,nroi,ntype]
    I_stw_ij_all = np.zeros((npos,npos,nsubj,len(roi_labels),5))*np.nan # [npos(s1),npos(s2),nsubj,nroi,ntype]
    Ibc_stw_ij_all = np.zeros((npos,npos,nsubj,len(roi_labels),5))*np.nan # [npos(s1),npos(s2),nsubj,nroi,ntype]
    for roi_i in range(nroi):
        for subj_i in range(nsubj):
            stimposicond_unatt = stimposicond_all[1,subj_i,:]
            stimposicond_att = stimposicond_all[0,subj_i,:]
            # beta for each position (trial-by-trial)
            beta_digit_all = np.zeros((nvoxel,ntrialperpos,npos))*np.nan # voxel x trial x position
            beta_digit_mean_all = np.zeros((nvoxel,npos))*np.nan
            beta_face_all = np.zeros((nvoxel,ntrialperpos,npos))*np.nan # voxel x trial x position
            beta_face_mean_all = np.zeros((nvoxel,npos))*np.nan
            beta_digit_cov_all = np.zeros((nvoxel,nvoxel,npos))*np.nan # voxel x voxel x position
            beta_face_cov_all = np.zeros((nvoxel,nvoxel,npos))*np.nan
            for pos in range(npos):
                beta_digit_all[:,:,pos] = beta_all[1,roi_i,subj_i,:,stimposicond_unatt==pos + 1].T
                beta_digit_mean_all[:,pos] = np.nanmean(beta_digit_all[:,:,pos],axis=1)
                beta_face_all[:,:,pos] = beta_all[0,roi_i,subj_i,:,stimposicond_att==pos + 1].T
                beta_face_mean_all[:,pos] = np.nanmean(beta_face_all[:,:,pos],axis=1)
                beta_digit_cov_all[:,:,pos] = np.cov(beta_digit_all[:,:,pos])
                beta_face_cov_all[:,:,pos] = np.cov(beta_face_all[:,:,pos])
            for s1 in range(npos-1):
                for s2 in range(s1+1,npos):
                    ds,df_digit,dfnorm_digit,v_digit,ullambda_digit,htlambda_digit = cal_lfi_param_cov(beta_digit_mean_all,beta_digit_cov_all,s1,s2)
                    ds,df_face,dfnorm_face,v_face,ullambda_face,htlambda_face = cal_lfi_param_cov(beta_face_mean_all,beta_face_cov_all,s1,s2)
                    # I_digit   I_se    I_ms I_sr,mw  I_face

                    I_stw_ij_all[[s1,s2],[s2,s1],subj_i,roi_i,0] = cal_lfi(ds,dfnorm_digit,ullambda_digit,df_digit/dfnorm_digit,v_digit,htlambda_digit,False,ntrialperpos, nvoxel)
                    I_stw_ij_all[[s1,s2],[s2,s1],subj_i,roi_i,1] = cal_lfi(ds,dfnorm_face,ullambda_digit,df_digit/dfnorm_digit,v_digit,htlambda_digit,False,ntrialperpos, nvoxel)
                    I_stw_ij_all[[s1,s2],[s2,s1],subj_i,roi_i,2] = cal_lfi(ds,dfnorm_digit,ullambda_face,df_digit/dfnorm_digit,v_digit,htlambda_digit,False,ntrialperpos, nvoxel)
                    I_stw_ij_all[[s1,s2],[s2,s1],subj_i,roi_i,3] = cal_lfi(ds,dfnorm_digit,ullambda_digit,df_face/dfnorm_face,v_face,htlambda_face,False,ntrialperpos, nvoxel)
                    I_stw_ij_all[[s1,s2],[s2,s1],subj_i,roi_i,4] = cal_lfi(ds,dfnorm_face,ullambda_face,df_face/dfnorm_face,v_face,htlambda_face,False,ntrialperpos, nvoxel)

                    Ibc_stw_ij_all[[s1,s2],[s2,s1],subj_i,roi_i,:] = cal_lfi_bc(I_stw_ij_all[[s1,s2],[s2,s1],subj_i,roi_i,:], ntrialperpos, nvoxel, ds)

    Ilog_stw_ij_all = np.log(I_stw_ij_all)

    deltaIlog_stw_ij_all = Ilog_stw_ij_all[...,1:]-Ilog_stw_ij_all[...,[0]]

    Ilog_stw_d_subj = np.zeros((len(d),nsubj,len(roi_labels),5))*np.nan # d x subj x roi x task
    for dd in range(len(d)):
        Ilog_stw_d_subj[dd,:,:,:] = np.nanmean(Ilog_stw_ij_all[ds_list==d[dd],:,:,:],axis=0)
    deltaIlog_stw_d_subj = Ilog_stw_d_subj[:,:,:,1:]-Ilog_stw_d_subj[:,:,:,[0]]# d x subj x roi x type

    I_stw_d_subj = np.zeros((len(d),nsubj,len(roi_labels),5))*np.nan # d x subj x roi x task
    for dd in range(len(d)):
        I_stw_d_subj[dd,:,:,:] = np.nanmean(I_stw_ij_all[ds_list==d[dd],:,:,:],axis=0)

    Ibc_stw_d_subj = np.zeros((len(d),nsubj,len(roi_labels),5))*np.nan # d x subj x roi x task
    for dd in range(len(d)):
        Ibc_stw_d_subj[dd,:,:,:] = np.nanmean(Ibc_stw_ij_all[ds_list==d[dd],:,:,:],axis=0)

    return Ilog_stw_ij_all,deltaIlog_stw_ij_all,I_stw_ij_all,Ibc_stw_ij_all,Ilog_stw_d_subj,deltaIlog_stw_d_subj,I_stw_d_subj,Ibc_stw_d_subj


def run_simanalysis_lfi_oneexp(beta_all, stimposicond_all, ntrialperpos_sim, seed=None,):
    
    #print(f"Simulating data (seed={seed}) ...")
    rng = np.random.default_rng(seed)
    # simulate data
    beta_sim_all,stimposicond_sim_all = simulate_data(beta_all, stimposicond_all, ntrialperpos_sim, rng=rng)

    #print('Analyzing data ...')
    # lfi, mechanisms
    _,_,_,_,Ilog_stw_d_subj,deltaIlog_stw_d_subj,I_stw_d_subj,Ibc_stw_d_subj = lfi_mechanism_task(beta_sim_all,stimposicond_sim_all)
    return {"Ilog_stw_d_subj":Ilog_stw_d_subj,
            "deltaIlog_stw_d_subj":deltaIlog_stw_d_subj,
            "I_stw_d_subj":I_stw_d_subj,
            "Ibc_stw_d_subj":Ibc_stw_d_subj}

def run_simanalysis_lfi(beta_all,stimposicond_all,ntrialperpos_sim,n_exp=200,n_jobs=None,seed=0,save=True,savefilename='../expectedoutput/trialest_lfi.npz'):
    print(f"Running {n_exp} simulations in parallel (threading) ...")

    if n_jobs is None:
        n_jobs = min(32, n_exp)

    seed_sequence = np.random.SeedSequence(seed)
    child_seeds = seed_sequence.spawn(n_exp)
    exp_seeds = [int(s.generate_state(1)[0]) for s in child_seeds]

    results = [None] * n_exp

    with ThreadPoolExecutor(max_workers=n_jobs) as executor:
        future_to_idx = {
            executor.submit(
                run_simanalysis_lfi_oneexp,
                beta_all,
                stimposicond_all,
                ntrialperpos_sim,
                exp_seeds[i],
            ): i
            for i in range(n_exp)
        }

        for future in as_completed(future_to_idx):
            i = future_to_idx[future]
            results[i] = future.result()
            print(f"Finished simulation {i + 1}/{n_exp}")

    output = {
        key: np.stack([result[key] for result in results], axis=0)
        for key in results[0]
    }

    if save:
        np.savez(savefilename, **output)
        print(f"Saved results to: {savefilename}")

    return output
```

```python
from joblib import Parallel, delayed

def cal_dfnorm(beta_all,stimposicond_all):
    '''
    return: [npos,npos,nsubj,nroi,ntask]
    '''

    ntask,nroi,nsubj,nvoxel,_ = beta_all.shape
    npos = len(np.unique(stimposicond_all[~np.isnan(stimposicond_all)]))
    dfnorm_ij_all = np.zeros((npos,npos,nsubj,nroi,ntask)) # [npos(s1),npos(s2),nsubj,nroi,ntask]
    for task_i in range(ntask):
        for roi_i in range(nroi):
            for subj_i in range(nsubj):
                stimposicond = stimposicond_all[task_i,subj_i]
                beta_mean = np.zeros((nvoxel,npos))*np.nan
                for pos in range(npos):
                    beta_mean[:,pos] = np.nanmean(beta_all[task_i,roi_i,subj_i,:,stimposicond==pos].T,axis=1)
                for s1 in range(npos):
                    dfnorm_ij_all[s1,:,subj_i,roi_i,task_i] = np.linalg.norm(beta_mean[:,s1] - beta_mean.T,axis=1)
    
    return dfnorm_ij_all

def cal_ullambda(beta_all,stimposicond_all):
    '''
    return: [npos,npos,nsubj,nroi,ntask]
    '''
    ntask,nroi,nsubj,nvoxel,_ = beta_all.shape
    npos = len(np.unique(stimposicond_all[~np.isnan(stimposicond_all)]))
    ullambda_ij_all = np.zeros((npos,npos,nsubj,nroi,ntask)) # [npos(s1),npos(s2),nsubj,nroi,ntask]
    for task_i in range(ntask):
        for roi_i in range(nroi):
            for subj_i in range(nsubj):
                stimposicond = stimposicond_all[task_i,subj_i]
                beta = beta_all[task_i,roi_i,subj_i,:]
                for s1 in range(npos):
                    for s2 in range(s1, npos):
                        var1 = np.nanvar(beta[:,stimposicond==s1].T,axis=1,ddof=1)
                        var2 = np.nanvar(beta[:,stimposicond==s2].T,axis=1,ddof=1)
                        ullambda_ij_all[[s1,s2],[s2,s1],subj_i,roi_i,task_i] = np.nanmean((var1+var2)/2)
    
    return ullambda_ij_all


def cal_srdw(beta_all,stimposicond_all):
    '''
    input: 
        - beta_all: [task, roi, subj, voxel, trial]
        - stimposicond_all: [task, subj, trial]
    output:
        - vals_ij_all: [npos,npos,nsubj,nroi,ntask]
    '''
    ntask, nroi, nsubj, nvoxel, _ = beta_all.shape
    npos = len(np.unique(stimposicond_all[~np.isnan(stimposicond_all)]))
    ntrialperpos = np.max([np.sum(stimposicond_all[task_i,subj_i,:]==pos) for pos in range(npos) for task_i in range(ntask) for subj_i in range(nsubj)])

    vals_ij_all = np.zeros((npos,npos,nsubj,nroi,ntask)) # [npos(s1),npos(s2),nsubj,nroi,ntask]
    for task_i in range(ntask):
        for roi_i in range(nroi):
            for subj_i in range(nsubj):
                stimposicond = stimposicond_all[task_i,subj_i,:]
                beta_cond = np.zeros((nvoxel,ntrialperpos,npos))*np.nan 
                beta_mean_cond = np.zeros((nvoxel,npos))*np.nan
                for pos in range(npos):
                    b = beta_all[task_i,roi_i,subj_i,:,stimposicond==pos].T
                    beta_cond[:,:b.shape[1],pos] = b
                    beta_mean_cond[:,pos] = np.nanmean(beta_cond[:,:,pos],axis=1)
                for s1 in range(npos-1):
                    for s2 in range(s1+1,npos):
                        _,df,dfnorm,v,_,htlambda = cal_lfi_param(beta_mean_cond,beta_cond,s1,s2)
                        dfunit = df/dfnorm
                        vals_ij_all[[s1,s2],[s2,s1],subj_i,roi_i,task_i] = np.nansum([(dfunit@v[:,i])**2/htlambda[i] for i in range(len(htlambda))])

    return vals_ij_all

def cal_lfi_param(beta_mean_cond,beta_cond,s1,s2):
    # ds
    dr = s1//4-s2//4
    dc = s1%4-s2%4
    ds = ds0 *  (dr**2+dc**2)**0.5
    # df & |dfnorm|
    df = np.mat(beta_mean_cond[:,s1] - beta_mean_cond[:,s2])
    dfnorm = np.linalg.norm(df,axis=1)
    # eigenvector & eigenvalue
    beta_cond_s1 = beta_cond[:,:,s1]
    beta_cond_s2 = beta_cond[:,:,s2]
    beta_cond_s1 = beta_cond_s1[:,np.all(np.isfinite(beta_cond_s1),axis=0)]
    beta_cond_s2 = beta_cond_s2[:,np.all(np.isfinite(beta_cond_s2),axis=0)]
    q1 = np.cov(beta_cond_s1)
    q2 = np.cov(beta_cond_s2)
    q = (q1 + q2)/2
    w,v = np.linalg.eigh(q)
    ullambda = np.nanmean(w)
    htlambda = w/ullambda
    return ds,df,dfnorm,v,ullambda,htlambda


def srdw_leaveoneout(beta_session,stim_session):
    npos = len(np.unique(stim_session[~np.isnan(stim_session)]))
    ntask, nroi, nsubj, nsession,nvoxel,ntrial = beta_session.shape
    vals_ij_all = np.full([npos,npos,nsubj,nroi,ntask,nsession],np.nan)
    for ss in tqdm(range(nsession)):
        beta_tmp = beta_session.transpose(0,1,2,4,5,3).copy()
        beta_tmp[...,ss] = np.nan
        stim_tmp = stim_session.transpose(0,1,3,2).copy()
        stim_tmp[...,ss] = np.nan
        beta_tmp = beta_tmp.reshape([ntask,nroi,nsubj,nvoxel,nsession*ntrial])
        stim_tmp = stim_tmp.reshape([ntask,nsubj,nsession*ntrial])
        vals_ij_all[...,ss] = cal_srdw(beta_tmp,stim_tmp)

    return vals_ij_all
        

def metric_distance(metric_ij_all):
    '''
    input: metric_ij_all [npos,npos,nsubj,nroi,ntask,...]
    output: metric_d_subj [d,nsubj,nroi,ntask,...]
    '''
    # average dfnorm across position pairs for each distance for each subject
    metric_d_subj = np.full([len(d)]+list(metric_ij_all.shape[2:]),np.nan) # d x subj x roi x task x ...
    for dd in range(len(d)):
        metric_d_subj[dd,:,:,:] = np.nanmean(metric_ij_all[ds_list==d[dd],:,:,:],axis=0)
    
    return metric_d_subj    

```

```python
ntask,nroi,nsubj,nsession,nvoxel,_ = beta_session.shape

# |df|
dfnorm_ij_session = np.full([npos,npos,nsubj,nroi,ntask,nsession],np.nan)
for session_i in range(nsession):
    dfnorm_ij_session[...,session_i] = cal_dfnorm(beta_session[:,:,:,session_i,:,:],stim_session[:,:,session_i,:])

logdfnorm_ij_session = np.log(dfnorm_ij_session)
logdfnorm_ij_session[np.isinf(logdfnorm_ij_session)] = np.nan

logdfnorm_d_session = metric_distance(logdfnorm_ij_session) # d x subj x roi x task x session

# <lambda>
ullambda_ij_session = np.full([npos,npos,nsubj,nroi,ntask,nsession],np.nan)
for session_i in range(nsession):
    ullambda_ij_session[...,session_i] = cal_ullambda(beta_session[:,:,:,session_i,:,:],stim_session[:,:,session_i,:])

logullambda_ij_session = np.log(ullambda_ij_session)
logullambda_ij_session[np.isinf(logullambda_ij_session)] = np.nan

logullambda_d_session = metric_distance(logullambda_ij_session)


# <dfunit,w,v> leave-one-out
vals_ij_session = srdw_leaveoneout(beta_session,stim_session)
# [npos,npos,nsubj,nroi,ntask,nsession]

logvals_ij_session = np.log(vals_ij_session)
logvals_ij_session[np.isinf(logvals_ij_session)] = np.nan

logvals_d_session = metric_distance(logvals_ij_session)


np.savez('../expectedoutput/driftcheck_session_loo.npz',
    logdfnorm_d_session = logdfnorm_d_session,
    logullambda_d_session = logullambda_d_session,
    logvals_d_session = logvals_d_session,
    )

```
