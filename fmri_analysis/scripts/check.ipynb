```python
import numpy as np

def npz_equal(path1, path2):
    d1 = np.load(path1)
    d2 = np.load(path2)

    # 1. 键名集合是否相同
    if set(d1.files) != set(d2.files):
        print("键名不同")
        print("只在 a 中:", set(d1.files) - set(d2.files))
        print("只在 b 中:", set(d2.files) - set(d1.files))
        return False

    # 2. 逐个数组比较
    for k in d1.files:
        a1 = d1[k]
        a2 = d2[k]
        if a1.shape != a2.shape:
            print(f"{k}: shape 不同 {a1.shape} vs {a2.shape}")
            return False
        if a1.dtype != a2.dtype:
            print(f"{k}: dtype 不同 {a1.dtype} vs {a2.dtype}")
            return False
        if not np.array_equal(a1, a2):
            print(f"{k}: 值不同")
            print(a1[0,0])
            print(a2[0,0])

            return False

    return True

#print(npz_equal('../../eye_analysis/eye_results/contour_areas_session.npz', '../../../o/eye_analysis/eye_results/contour_areas_session.npz'))
print(npz_equal('../expectedoutput/svmacc.npz', '../../plot_figures/data/svmacc.npz'))
```
